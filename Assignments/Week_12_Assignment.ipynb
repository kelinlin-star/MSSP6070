```python
import pandas as pd
import scipy.stats as stats
```

```python
!git clone https://github.com/kelinlin-star/MSSP6070.git
%cd MSSP6070/data
```

```python
df = pd.read_csv("Student Performance Data (1).csv")
```

```python
group1 = df[(df['gender'] == 'male') &
            (df['test preparation course'] == 'completed')]['total score']

group2 = df[~((df['gender'] == 'male') &
              (df['test preparation course'] == 'completed'))]['total score']

t_stat, p_val = stats.ttest_ind(group1, group2, equal_var=False, nan_policy='omit')

print("T statistic:", t_stat)
print("P value:", p_val)

if p_val < 0.05:
    print("Conclusion: Males who completed the test preparation course scored significantly higher than other students.")
else:
    print("Conclusion: There is no significant difference between males who completed the preparation course and other students.")
```

```python
male = df[df['gender'] == 'male']['total score']
female = df[df['gender'] == 'female']['total score']

t_stat, p_val = stats.ttest_ind(male, female, equal_var=False)
print("Male vs Female - p value:", p_val)

if p_val < 0.05:
    print("Conclusion: There is a significant difference between male and female students, and females tend to score higher overall.")
else:
    print("Conclusion: There is no significant difference between male and female students in overall performance.")
```
