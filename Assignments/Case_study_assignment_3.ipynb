```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

```python
data = pd.read_csv("HousingPrices.csv")
```

```python
mean_price = data['SalePrice'].mean()
std_price = data['SalePrice'].std()
print(f"Mean Price: {mean_price}")
print(f"Standard Deviation of Price: {std_price}")
```

```python
plt.figure(figsize=(10, 6))
plt.hist(data['SalePrice'], bins=50, color='blue', edgecolor='black')
plt.title('Distribution of Housing Prices')
plt.xlabel('Sale Price')
plt.ylabel('Frequency')
plt.grid(True)
plt.show()
```

```python
Q1 = data['SalePrice'].quantile(0.25)
Q3 = data['SalePrice'].quantile(0.75)
IQR = Q3 - Q1
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR
outliers = data[(data['SalePrice'] < lower_bound) | (data['SalePrice'] > upper_bound)]
print(f"Number of Outliers: {outliers.shape[0]}")
```
