```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
from scipy import stats
from sklearn.preprocessing import LabelEncoder
```

```python
!git clone https://github.com/kelinlin-star/MSSP6070.git
%cd MSSP6070/data
```

```python
data = pd.read_csv("student_depression_dataset.csv")
data.head()
```

```python
# Convert categorical columns to numeric
label_encoder = LabelEncoder()

# List of categorical columns to encode
categorical_columns = ['Gender', 'City', 'Profession', 'Sleep Duration', 'Dietary Habits', 'Degree',
                       'Have you ever had suicidal thoughts ?', 'Financial Stress', 'Family History of Mental Illness']

# Apply label encoding to each categorical column
for col in categorical_columns:
    data[col] = label_encoder.fit_transform(data[col])
```

```python
# Check for missing values
missing_values = data.isnull().sum()
print("Missing values:\n", missing_values)
```

```python
# Descriptive statistics (mean, standard deviation)
mean_values = data.mean()
std_values = data.std()

print("Mean values:\n", mean_values)
print("Standard Deviation values:\n", std_values)
```

```python
# Visualize the distributions of numeric columns
plt.figure(figsize=(10, 8))
data.hist(figsize=(10, 8))
plt.suptitle('Histograms of Numeric Variables', fontsize=16)
plt.tight_layout()
plt.show()
```

```python
# Boxplot for identifying outliers
numerical_data = data.select_dtypes(include=[np.number])

plt.figure(figsize=(15, 12))
for i, col in enumerate(numerical_data.columns, 1):
    plt.subplot(5, 4, i)  # 5x4 grid for 17 variables
    sns.boxplot(data=numerical_data[col])
    plt.title(f'{col}', fontsize=12)
    plt.xticks(rotation=45, fontsize=10)
    plt.yticks(fontsize=10)

plt.tight_layout()
plt.show()
```

```python
# Check for normality of a numeric variable (e.g., CGPA)
plt.figure(figsize=(10, 8))
sns.histplot(data['CGPA'], kde=True)
plt.title('CGPA Distribution with KDE', fontsize=16)
plt.show()
```

```python
# Shapiro-Wilk test for normality on the 'CGPA' column
shapiro_test_cgpa = stats.shapiro(data['CGPA'])
print(f"\nShapiro-Wilk test p-value for CGPA: {shapiro_test_cgpa[1]}")
```

```python
# Outliers Detection using Z-scores
z_scores = np.abs(stats.zscore(data['CGPA']))
outliers = data[z_scores > 3]
print(f"\nOutliers based on Z-score for CGPA:\n{outliers}")
```

```python
# Correlation matrix to identify relationships between numerical variables
correlation_matrix = data.corr()
plt.figure(figsize=(12, 10))
sns.heatmap(correlation_matrix, annot=True, fmt=".2f", cmap='coolwarm', square=True, linewidths=0.5, cbar_kws={"shrink": .7})
plt.title('Correlation Matrix', fontsize=16)
plt.tight_layout()
plt.show()
```

```python
print("\n--- Summary ---")
print(f"Missing Values in Data: \n{missing_values}")
print(f"\nMean Values: \n{mean_values}")
print(f"\nStandard Deviation: \n{std_values}")
print(f"\nShapiro-Wilk Test for Normality (CGPA): p-value = {shapiro_test_cgpa[1]}")
print(f"\nOutliers in CGPA (Z-score > 3): \n{outliers}")
print(f"\nCorrelation Matrix: \n{correlation_matrix}")
```
